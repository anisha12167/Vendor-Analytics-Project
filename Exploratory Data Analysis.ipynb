## Data Overview

The dataset contains multiple tables storing vendor purchase, pricing, invoice, and sales information.

- **Purchases Table:** Contains vendor purchase transactions including date, brand, quantity, and amount paid.  
- **Purchase Prices Table:** Provides actual product and purchase prices with a unique vendor–brand combination.  
- **Vendor Invoice Table:** Aggregates purchase data with total quantity, purchase amount, and freight cost, unique by vendor and PO number.  
- **Sales Table:** Records sales transactions including brand, quantity sold, selling price, and revenue.

## Summary Table Requirement

Since the required analysis data is distributed across multiple tables, we need to create a consolidated summary table that combines:
- Vendor purchase transactions  
- Sales transaction details  
- Vendor freight costs  
- Actual product pricing  

Next, we will explore each table individually before building the summary table.


```python
freight_summary = pd.read_sql_query("""SELECT 
    VendorNumber, 
    SUM(Freight) as FreightCost 
FROM vendor_invoice 
GROUP BY VendorNumber""", conn)
freight_summary
```

```python
purchase_summary = pd.read_sql_query("""SELECT 
    p.VendorNumber,
    p.VendorName,
    p.Brand,
    p.Description,
    p.PurchasePrice,
    pp.Price as ActualPrice,
    pp.Volume,
    SUM(p.Quantity) AS TotalPurchaseQuantity,
    SUM(p.Dollars) AS TotalPurchaseDollars
FROM purchases p
JOIN purchase_prices pp
    ON p.Brand = pp.Brand
WHERE p.PurchasePrice > 0
GROUP BY p.VendorNumber,p.VendorName, p.Brand, p.Description, p.PurchasePrice, pp.Price, pp.Volume
ORDER BY TotalPurchaseDollars
""",conn)
purchase_summary
```

```python
sales_summary = pd.read_sql_query("""
SELECT VendorNo, Brand,
SUM(SalesQuantity) AS TotalSalesQuantity,
SUM(SalesDollars) AS TotalSalesDollars
FROM sales
GROUP BY VendorNo, Brand
LIMIT 10000
""", conn)
```

```python
vendor_sales_summary = purchase_summary.merge(
    sales_summary,
    left_on=["VendorNumber", "Brand"],
    right_on=["VendorNo", "Brand"],
    how="left"
)
```

```python
vendor_sales_summary.to_sql(
    "vendor_sales_summary",
    conn,
    if_exists="replace",
    index=False
)
```

## Data Cleaning

Data cleaning ensures the `vendor_sales_summary` table is accurate, consistent, and reliable for analysis and reporting.

### Key Steps
- **Handle Missing Values:** Remove or fill null values in important columns.
- **Remove Duplicates:** Prevent incorrect aggregation and analysis.
- **Standardize Text Fields:** Ensure consistent vendor, brand, and product names.
- **Validate Numerical Data:** Check for negative or invalid values in sales, pricing, and freight.
- **Fix Data Types:** Maintain correct formats for numeric, text, and date columns.

### Outcome
Clean data improves analysis accuracy, speeds up dashboard performance, and supports better vendor and pricing decisions.

1. The Volume is a numerical column but having object datatype.
2. There are some products that are not sold, so having missing values.
3. There are white spaces in the categorical columns.

```python
# creating new columns for better analysis
vendor_sales_summary['GrossProfit'] = vendor_sales_summary['TotalSalesDollars'] - vendor_sales_summary['TotalPurchaseDollars']
vendor_sales_summary['ProfitMargin'] = (vendor_sales_summary['GrossProfit'] / vendor_sales_summary['TotalSalesDollars'])*100
vendor_sales_summary['StockTurnover'] = vendor_sales_summary['TotalSalesQuantity'] / vendor_sales_summary['TotalPurchaseQuantity']
vendor_sales_summary['SalesToPurchaseRatio'] = vendor_sales_summary['TotalSalesDollars'] / vendor_sales_summary['TotalPurchaseDollars']
```

```python
# changing datatype to float
vendor_sales_summary['Volume'] = vendor_sales_summary['Volume'].astype('float')

# filling missing value with 0
vendor_sales_summary.fillna(0,inplace = True)

# removing spaces from categorical columns
vendor_sales_summary['VendorName'] = vendor_sales_summary['VendorName'].str.strip()
vendor_sales_summary['Description'] = vendor_sales_summary['Description'].str.strip()
```

### Saving Cleaned Data into Database

After completing data cleaning, the refined dataset is stored back into the database. This ensures that dashboards, reports, and analytics processes use accurate, consistent, and optimized data, reducing query time and improving overall performance.


```python
cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS vendor_sales_summary (
    VendorNumber INT,
    VendorName VARCHAR(100),
    Brand INT,
    Description VARCHAR(100),
    PurchasePrice DECIMAL(10,2),
    ActualPrice DECIMAL(10,2),
    Volume INT,
    TotalPurchaseQuantity INT,
    TotalPurchaseDollars DECIMAL(15,2),
    TotalSalesQuantity INT,
    TotalSalesDollars DECIMAL(15,2),
    TotalSalesPrice DECIMAL(15,2),
    TotalExciseTax DECIMAL(15,2),
    FreightCost DECIMAL(15,2),
    GrossProfit DECIMAL(15,2),
    ProfitMargin DECIMAL(15,2),
    StockTurnover DECIMAL(15,2),
    SalesToPurchaseRatio DECIMAL(15,2),
    PRIMARY KEY (VendorNumber, Brand)
);
""")
```
